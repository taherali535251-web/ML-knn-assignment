```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

import warnings
warnings.filterwarnings("ignore")
```

```python
#seaborn: statistical data visualization
sns.get_dataset_names()
```

```python
df = sns.load_dataset('tips')
df
```

```python
#numerical data plotting
sns.relplot(x = 'total_bill',y = 'tip',data = df) #the relationship between two variables, by default it gives scatterplot, kind='line'
```

```python
sns.scatterplot(x= df.total_bill, y=df.tip, data = df)
```

```python
sns.scatterplot(x= df.total_bill, y=df.tip, data = df, hue = 'smoker')
```

```python
# as the bil increase tip increses highest tip was paid by smoker
sns.relplot(x = 'total_bill',y = 'tip',data = df,hue ='smoker')
```

```python
sns.relplot(x = 'total_bill',y = 'tip',data = df,hue = 'size')
```

```python
sns.relplot(x = 'total_bill',y = 'tip',data = df,hue = 'time', style = 'smoker')
```

```python
value = np.random.randn(500).cumsum()
time=np.arange(500)
df = pd.DataFrame({'time': time,'value':value})
df
```

```python
#relpolot to make line plot
sns.relplot(x = 'time', y = 'value',kind ='line',data= df,sort = True)
```
