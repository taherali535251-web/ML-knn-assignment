```python
#5 Create a list and tuple in Python, and demonstrate how attempting to change an element works differently for each.
# Creating a list and a tuple
my_list = [10, 20, 30]
my_tuple = (10, 20, 30)

# ✅ Lists are mutable
print("Original list:", my_list)
my_list[1] = 99   # Changing the second element
print("Modified list:", my_list)

# ❌ Tuples are immutable
print("Original tuple:", my_tuple)
try:
    my_tuple[1] = 99   # Attempting to change the second element
except TypeError as e:
    print("Error when modifying tuple:", e)
```

```python
#6. Write a function to demonstrate the behavior of mutable and immutable arguments
# Function to modify arguments
def modify_data(mutable_arg, immutable_arg):
    # Modify the mutable argument (list)
    mutable_arg.append(99)

    # Try to modify the immutable argument (int)
    immutable_arg += 10

    print("Inside function:")
    print("Mutable argument:", mutable_arg)
    print("Immutable argument:", immutable_arg)


# Main program
my_list = [1, 2, 3]   # mutable
my_number = 5         # immutable

print("Before function call:")
print("List:", my_list)
print("Number:", my_number)

modify_data(my_list, my_number)

print("\nAfter function call:")
print("List:", my_list)
print("Number:", my_number)

modify_data(my_list, my_number)

print("\nAfter function call:")
print("List:", my_list)       # Changed because lists are mutable
print("Number:", my_number)   # Unchanged because integers are immutable
```

```python
#7. Write a program that performs basic arithmetic operations on two user-input numbers.
# Program to perform basic arithmetic operations

# Taking user input
num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))

# Performing operations
print("\nResults:")
print("Addition:", num1 + num2)
print("Subtraction:", num1 - num2)
print("Multiplication:", num1 * num2)

# Division with check to avoid ZeroDivisionError
if num2 != 0:
    print("Division:", num1 / num2)
else:
    print("Division: Error! Cannot divide by zero.")
```

```python
#8. Write a program to demonstrate the use of logical operators
# Program to demonstrate logical operators

# Taking user input
num1 = int(input("Enter first number: "))
num2 = int(input("Enter second number: "))

print("\nResults:")

# Using 'and' operator
if num1 > 0 and num2 > 0:
    print("Both numbers are positive.")

# Using 'or' operator
if num1 % 2 == 0 or num2 % 2 == 0:
    print("At least one number is even.")

# Using 'not' operator
if not (num1 == num2):
    print("The numbers are not equal.")
```

```python
#9. Write a Python program to convert user input from string to integer, float, and boolean types.
# Program to convert user input into different data types

# Taking input from user (always comes as string)
user_input = input("Enter something: ")

print("\nOriginal input (string):", user_input)

# Convert to integer
try:
    int_value = int(user_input)
    print("As integer:", int_value)
except ValueError:
    print("Cannot convert to integer.")

# Convert to float
try:
    float_value = float(user_input)
    print("As float:", float_value)
except ValueError:
    print("Cannot convert to float.")

# Convert to boolean
# In Python, bool() returns False for empty string "" and "0", True otherwise
bool_value = bool(user_input)
print("As boolean:", bool_value)
```

```python
#10. Write code to demonstrate type casting with list elements
# Program to demonstrate type casting with list elements

# Original list of strings
data = ["10", "20.5", "True", "hello"]

print("Original list:", data)

# Convert elements to different types
int_value = int(data[0])        # string "10" → integer 10
float_value = float(data[1])    # string "20.5" → float 20.5
bool_value = bool(data[2])      # string "True" → boolean True (non-empty string → True)
str_value = str(data[3])        # string "hello" → remains string

# Display results
print("\nAfter type casting:")
print("Integer:", int_value)
print("Float:", float_value)
print("Boolean:", bool_value)
print("String:", str_value)
```

```python
#13 Write a Python program to find the sum of all even numbers between 1 and 50
# Program to find the sum of all even numbers between 1 and 50

total = 0

for num in range(2, 51, 2):   # start at 2, step by 2 → only even numbers
    total += num

print("The sum of all even numbers between 1 and 50 is:", total)
```

```python
#14. Write a program to reverse a string using a while loop.
# Program to reverse a string using a while loop

# Take user input
text = input("Enter a string: ")

# Initialize variables
reversed_text = ""
index = len(text) - 1   # Start from the last character

# Loop until the beginning of the string
while index >= 0:
    reversed_text += text[index]
    index -= 1

print("Reversed string:", reversed_text)
```

```python
#15. Write a Python program to calculate the factorial of a number provided by the user using a while loop
# Program to calculate factorial using while loop

# Taking user input
num = int(input("Enter a number: "))

# Initialize variables
factorial = 1
i = 1

# While loop to calculate factorial
while i <= num:
    factorial *= i
    i += 1

print(f"The factorial of {num} is {factorial}")
```
