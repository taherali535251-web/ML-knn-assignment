```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
from sklearn.datasets import load_iris
```

```python
data = load_iris()
```

```python
print(data.feature_names)
```

```python
print(data.DESCR)
```

```python
df = pd.DataFrame(data.data, columns= data.feature_names)
```

```python
df['target'] = data.target
```

```python
df
```

```python
x = df.drop('target',axis=1)
```

```python
y= df['target']
```

```python
x
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.3, random_state=1)
```

```python
x_train.shape
```

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
d = DecisionTreeClassifier(criterion='entropy')
```

```python
d.fit(x_train,y_train)
```

```python
from sklearn import tree
plt.Figure(figsize=(12,10))
tree.plot_tree(d,filled=True)
```

```python
d = DecisionTreeClassifier(criterion='entropy', max_depth= 2)
```

```python
from sklearn import tree
plt.Figure(figsize=(12,10))
tree.plot_tree(d,filled=True)
```

```python
y_pred = d.predict(x_test)
```

```python
y_pred
```

```python
from sklearn.metrics import accuracy_score,classification_report
score = accuracy_score(y_pred,y_test)
print(score)
print(classification_report(y_pred,y_test))
```

```python
#preprunning using hyper parameter
params = {'criterion':['gini','entropy','log_loss'],
          'splitter':['best','random'],
          'max_depth':[1,2,3,4,5,6],
          'max_features':['sqrt','log2','auto']}
```

```python
from sklearn.model_selection import GridSearchCV
```

```python
clf = DecisionTreeClassifier()
model =GridSearchCV(clf,param_grid=params, cv= 5,scoring='accuracy')
```

```python
model.fit(x_train,y_train)
```

```python
model.best_params_
```

```python
model.best_score_
```

```python
y_pred = model.predict(x_test)
```

```python
score = accuracy_score(y_pred,y_test)
print(score)
print(classification_report(y_pred,y_test))
```

```python
from sklearn.datasets import fetch_california_housing
data = fetch_california_housing()
```

```python
df = pd.DataFrame(data.data, columns= data.feature_names)
df['price'] = data.target
```

```python
df
```

```python
x = df.drop('price',axis=1)
```

```python
y= df['price']
```

```python
df = df.sample(frac=0.20)
```

```python
df
```

```python
#independent features
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.20, random_state=1)
```

```python
x_train.shape
```

```python
from sklearn.tree import DecisionTreeRegressor
reg = DecisionTreeRegressor()
reg
```

```python
reg.fit(x_train,y_train)
```

```python
y_pred = reg.predict(x_test)
```

```python
from sklearn.metrics import r2_score
```

```python
r2_score(y_pred,y_test)
```

```python
#hyperparameter 
param = {
    'criterion':['squared_error','friedman_mse','absolute_error','poisson'],
    'splitter':['best','random'],
    'max_depth':[1,2,3,4,5,6],
    'max_features':['auto','sqrt','log2']}
regressor = DecisionTreeRegressor()
```
