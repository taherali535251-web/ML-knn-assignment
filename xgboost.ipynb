```python
from xgboost import XGBRegressor
from sklearn.datasets import make_regression
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score,mean_absolute_error,mean_squared_error
```

```python
x,y = make_regression(n_samples=1000,n_features=2,noise=10,random_state=1)
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.30,random_state=1)
```

```python
reg = XGBRegressor()
reg.fit(x_train,y_train)
```

```python
y_pred = reg.predict(x_test)
#evaluation the model
print("cutrrent model performance")
print(f"R2 score:{r2_score(y_test,y_pred)}")
print(f"MEAN ABSOLUTE ERROR:{mean_absolute_error(y_test,y_pred)}")
print(f"MEAN Squared error:{mean_squared_error(y_test,y_pred)}")
```

```python
#hyperparameter tuning
from sklearn.model_selection import GridSearchCV
param = {
    'n_estimators':[50, 100, 200],
    'learning_rate':[0.01, 0.1, 0.5, 1.0, 1.5]
    }
```

```python
greg = GridSearchCV(estimator=reg,param_grid=param,verbose=2,cv=5,n_jobs=-1)
```

```python
greg.fit(x_train,y_train)
```

```python
greg.best_params_
```

```python
best_model = greg.best_estimator_
y_pred = best_model.predict(x_test)

#evaluation the model
print("cutrrent model performance")
print(f"R2 score:{r2_score(y_test,y_pred)}")
print(f"MEAN ABSOLUTE ERROR:{mean_absolute_error(y_test,y_pred)}")
print(f"MEAN Squared error:{mean_squared_error(y_test,y_pred)}")
```

```python
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
from xgboost import XGBClassifier
from sklearn.metrics import accuracy_score,classification_report,confusion_matrix
```

```python
x,y = make_classification(n_samples=1000, n_features=20,n_classes=2,random_state=1)
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.30,random_state=1)
```

```python
cgb = XGBClassifier()
cgb.fit(x_train,y_train)
```

```python
y_pred = cgb.predict(x_test)

#evaluate model
print("current model performance")
print(f"accuracy:{accuracy_score(y_test,y_pred)}")
print(f"classification report:{classification_report(y_test,y_pred)}")
print(f"confusion_matrix:{confusion_matrix(y_test,y_pred)}")
```

```python
#hyperparametrer tuning
from sklearn.model_selection import GridSearchCV
param = {
    'n_estimators':[50, 100, 200],
    'learning_rate':[0.01, 0.1, 1.0, 1.5, 2.0]}
```

```python
grid = GridSearchCV(estimator=cgb,param_grid=param,cv=5,verbose=2,n_jobs=-1)
```

```python
grid.fit(x_train,y_train)
```

```python
grid.best_params_
```

```python
best_model = grid.best_estimator_
y_pred = best_model.predict(x_test)
print("current model performance")
print(f"accuracy:{accuracy_score(y_test,y_pred)}")
print(f"classification report:{classification_report(y_test,y_pred)}")
print(f"confusion_matrix:{confusion_matrix(y_test,y_pred)}")
```
